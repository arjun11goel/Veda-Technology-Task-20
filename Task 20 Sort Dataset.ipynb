# Task 20: Sort Dataset

## Description
Sort a dataset using numerical and categorical columns in ascending and descending order.

## Objective
Understand how sorting helps organize and analyze datasets.

## Tools Used
* Python
* Pandas
* Jupyter Notebook

```python
import pandas as pd

# Creating a sample Employee Dataset directly in the notebook
data = {
    'Employee_ID': [101, 102, 103, 104, 105, 106, 107, 108],
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Ethan', 'Fiona', 'George', 'Hannah'],
    'Department': ['IT', 'HR', 'IT', 'Finance', 'HR', 'Finance', 'IT', 'HR'],
    'Age': [28, 34, 25, 42, 30, 26, 38, 29],
    'Salary': [65000, 55000, 70000, 90000, 58000, 62000, 85000, 56000]
}

df = pd.DataFrame(data)

print("--- Original Employee Dataset ---")
print(df)
print("\n" + "="*50 + "\n")

# 1. Example of Ascending Sorting (Sorting by Age in ascending order - default)
print("1. Ascending Sorting (by Age - Lowest to Highest):")
sorted_asc = df.sort_values(by='Age')
print(sorted_asc)
print("\n" + "="*50 + "\n")

# 2. Example of Descending Sorting (Sorting by Salary in descending order)
print("2. Descending Sorting (by Salary - Highest to Lowest):")
sorted_desc = df.sort_values(by='Salary', ascending=False)
print(sorted_desc)
print("\n" + "="*50 + "\n")

# 3. Example of Sorting Using Multiple Columns
# Sort by Department alphabetically (ascending), then by Salary from highest to lowest (descending)
print("3. Multi-column Sorting (Department Ascending, Salary Descending):")
sorted_multi = df.sort_values(by=['Department', 'Salary'], ascending=[True, False])
print(sorted_multi)
```

## Interview Questions & Answers

### 1. How do you sort a DataFrame?
* **Answer:** You can sort a Pandas DataFrame using the `.sort_values()` method by specifying the column name(s) you want to sort by. 

### 2. What does ascending=False do?
* **Answer:** Setting `ascending=False` inside the `.sort_values()` method sorts the data in descending order (e.g., largest to smallest for numbers, or Z to A for text). By default, it is set to `True` (ascending order).

### 3. Can a DataFrame be sorted by multiple columns?
* **Answer:** Yes! You can sort by multiple columns by passing a list of column names to the `by` parameter (e.g., `df.sort_values(by=['Col1', 'Col2'])`). You can also control the ascending/descending order for each column individually by passing a corresponding list of booleans to the `ascending` parameter (e.g., `ascending=[True, False]`).
